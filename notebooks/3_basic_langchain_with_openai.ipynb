```python
from os import getenv
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_google_genai.chat_models import ChatGoogleGenerativeAIError
from langchain_openai import ChatOpenAI

prompt = "What is GenAI?"
google_api_key = getenv("GOOGLE_API_KEY")
openai_api_key = getenv("OPENAI_API_KEY")

if google_api_key:
    try:
        llm = ChatGoogleGenerativeAI(
            model="gemini-2.5-flash",
            google_api_key=google_api_key
        )
        response = llm.invoke(prompt)
        print(response.content)
    except ChatGoogleGenerativeAIError as err:
        print("Gemini API call failed. Falling back to OpenAI...")
        print(err)
        if not openai_api_key:
            raise ValueError("OPENAI_API_KEY is missing in .env.")
        llm = ChatOpenAI(model="gpt-5.6", api_key=openai_api_key)
        response = llm.invoke(prompt)
        print(response.content)
else:
    if not openai_api_key:
        raise ValueError("Set GOOGLE_API_KEY or OPENAI_API_KEY in .env.")
    llm = ChatOpenAI(model="gpt-5.6", api_key=openai_api_key)
    response = llm.invoke(prompt)
    print(response.content)
```
