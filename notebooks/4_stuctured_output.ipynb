```python
from dotenv import load_dotenv
import os
load_dotenv()

from langchain_openai import ChatOpenAI
llm = ChatOpenAI(model="gpt-5.6")


text = "Hello, My name is Amit Kumar Gupta, "\
"Email is amitkumar4268@gmail.com and my age is 40"

response = llm.invoke(f"Please give me the only name, email and age from this text: {text}")
print(response.content)
```

```python
from pydantic import BaseModel, Field
from typing import List, Optional

class ResponseStructire(BaseModel):
   name:str = Field(description="Name of the person")
   email:str = Field(description="Email of the person")
   age:int = Field(description="Age of the person")


structured_llm = llm.with_structured_output(ResponseStructire)
response = structured_llm.invoke(f"Please give me the only name, email and age from this text: {text}")
response.model_dump()
```

```python
class Movies(BaseModel):
    title:str = Field(description="Title of the movie")
    year:int = Field(description="Year of the movie released")

class MovieList(BaseModel):
    movies:List[Movies]


movie_llm = llm.with_structured_output(MovieList)

response = movie_llm.invoke("Give me a list of 5 movies with their title and year of release")
response.model_dump()
```
